```python
def quanqual(dataset):
    quan=[]
    qual=[]
    for columnname in dataset.columns:
        print (columnname)
        if (dataset[columnname].dtypes=="str"):
            print ("qual")
            qual.append (columnname)
        else:
            print ("quant")
            quan.append (columnname)
    return quan,qual
```

```python
quanqual(dataset)
```

```python
tablecreate=pd.DataFrame(index=["Mean","Median","Mode"],columns=quan)
for columnName in quan:
    tablecreate.loc["Mean",columnName]=dataset[columnName].mean()
    tablecreate.loc["Median",columnName]=dataset[columnName].median()
    tablecreate.loc["Mode",columnName]=dataset[columnName].mode()[0]
print(tablecreate)
```

```python
tablecreate=pd.DataFrame(index=["Mean","Median","Mode","Q1:25","Q2:50","Q3:75","Q4:100"],columns=quan)
for columnName in quan:
    tablecreate.loc["Mean",columnName]=dataset[columnName].mean()
    tablecreate.loc["Median",columnName]=dataset[columnName].median()
    tablecreate.loc["Mode",columnName]=dataset[columnName].mode()[0]
print(tablecreate)
```

```python
tablecreate=pd.DataFrame(index=["Mean","Median","Mode","Q1:25","Q2:50","Q3:75","Q4:100"],columns=quan)
for columnName in quan:
    tablecreate.loc[columnName]["Mean"]=dataset[columnname].mean()
    tablecreate.loc[columnName]["Median"]=dataset[columnname].median()
    tablecreate.loc[columnName]["Mode"]=dataset[columnname].mode()[0]
    tablecreate.loc[columnName]["Q1:25"]=np.percentile(dataset[columnname],60)
print(tablecreate)
```
